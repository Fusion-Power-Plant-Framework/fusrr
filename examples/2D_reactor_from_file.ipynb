```python
from renderingpipeline.reactor import Reactor
from renderingpipeline.views import View2D, View3D
from IPython.display import Image
PROCESS_file = "EUDEMO_MFILE.DAT"
BLUEMIRA_CAD = "EUDEMO.gltf"
```

# **Example View in Rendermak**
**This notebook demonstrates the usage of the rendermak module and provides an example of how to generate some views for visualising reactors**

This notebook is a work in progress, with the goal of displaying every view Rendermak can produce.


Importing the Reactor class from the reactor file, to create a reactor from the file name the user must call the `reactor_from_file` method and input the MFILE of choice. Every view needs a new Reactor object to render.

**2D VIEWS**

```python
reactor_2D_cross = Reactor.reactor_from_file(PROCESS_file)
```

Once the reactor has been created a view can be chosen to visualise the reactor. A  2D polodial cross section is set up.

```python
cross_section_2D = View2D(reactor_2D_cross)
cross_section_2D.half_reactor()
cross_section_2D.save_image("examplerenders/2D_cross_section.png")
cross_section_2D.reset_scene()
```

```python
Image(url="examplerenders/2D_cross_section.png")
```

To recreate another view, a new reactor must be instantiated and then called into a view class as before.

```python
reactor_key_2D = Reactor.reactor_from_file(PROCESS_file)
key_2D = View2D(reactor_key_2D)
key_2D.half_reactor()
key_2D.key()
key_2D.save_image("examplerenders/2D_cross_with_key.png")
key_2D.reset_scene()
```

```python
Image(url="examplerenders/2D_cross_with_key.png")
```

**3D VIEWS**

Reactors are made in 3D by default in their `build()` methods. Saving a reactor without call a view method will render the 3D reactor.

```python
reactor_3D = Reactor.reactor_from_file(PROCESS_file)
reactor_view_3D = View3D(reactor_3D)
reactor_view_3D.save_image("examplerenders/3D_reactor.png")
reactor_view_3D.reset_scene()
```

```python
Image(url="examplerenders/3D_reactor.png")
```

The 3D reactor can also have a key

```python
reactor_3D_key = Reactor.reactor_from_file(PROCESS_file)
reactor_view_3D_key = View3D(reactor_3D_key)
reactor_view_3D_key.key_3d()
reactor_view_3D_key.save_image("examplerenders/3D_reactor_with_key.png")
reactor_view_3D_key.reset_scene()
```

```python
Image(url="examplerenders/3D_reactor_with_key.png")
```
