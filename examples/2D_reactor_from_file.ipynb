# **Example View in the Rendering pipeline**
**This notebook demonstrates the usage of the rendering pipeline module and provides an example of how to generate some views for visualising reactors**

This notebook must be run in the directory where `MFILE.DAT` and `EUDEMO.gltf` test files live. 

Importing the Reactor class from the reactor file, to create a reactor from the file name the user must call the `reactor_from_file` method and input the MFILE of choice.

```python
from renderingpipeline.reactor import Reactor

example_reactor = Reactor.reactor_from_file("EUDEMO_MFILE.DAT")
```

Once the reactor has been created a view can be chosen to visualise the reactor. A default 2D polodial cross section is set up.

```python
from renderingpipeline.views import View
example_view = View(example_reactor)
example_reactor.save_image("example_view_reactor")
```
