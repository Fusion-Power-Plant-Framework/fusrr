```python
from fusrr.reactor import Reactor
from fusrr.views import View
from IPython.display import Image
import bpy
```

# **Example View in the Rendering pipeline**
**This notebook demonstrates the usage of the rendering pipeline module and provides an example of how to generate some views for visualising reactors**


Importing the Reactor class from the reactor file, to create a reactor from the file name the user must call the `reactor_from_file` method and input the MFILE of choice.

```python
example_reactor = Reactor.reactor_from_file("input-files/large_tokamak_MFILE.DAT", {
    "rpf1": "rpf[0]",
    "rpf2": "rpf[1]",
    "rpf3": "rpf[2]",
    "rpf4": "rpf[3]",
    "rpf5": "rpf[4]",
    "rpf6": "rpf[5]",
    "zpf1": "zpf[0]",
    "zpf2": "zpf[1]",
    "zpf3": "zpf[2]",
    "zpf4": "zpf[3]",
    "zpf5": "zpf[4]",
    "zpf6": "zpf[5]",
    "pfdr1": "pfdr(0)",
    "pfdr2": "pfdr(1)",
    "pfdr3": "pfdr(2)",
    "pfdr4": "pfdr(3)",
    "pfdr5": "pfdr(4)",
    "pfdr6": "pfdr(5)",
    "pfdz1": "pfdz(0)",
    "pfdz2": "pfdz(1)",
    "pfdz3": "pfdz(2)",
    "pfdz4": "pfdz(3)",
    "pfdz5": "pfdz(4)",
    "pfdz6": "pfdz(5)",
})
# example_reactor = Reactor.reactor_from_file("input-files/EUDEMO_MFILE.DAT")
```

Once the reactor has been created a view can be chosen to visualise the reactor. A default 2D polodial cross section is set up.

```python
example_view = View(example_reactor)
example_view.save_image("example_view_reactor")
from pathlib import Path
print(Path.cwd())
file_name = "blender-files/large-tokamak-single-null.blend"
filepath = str((Path.cwd() / file_name).resolve())
bpy.ops.wm.save_as_mainfile(filepath=filepath)
print(Path.cwd())
```
